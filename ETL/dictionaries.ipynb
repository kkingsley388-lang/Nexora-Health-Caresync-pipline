```python
# lets talk about nested dictionary, this is when we define a dictionary inside another dictionary, this is useful when we want to store more complex data structures. for example suppose we have a class and we want to define the marks of different students.
y = {
  "kelechi": { "maths": 90, "english": 80, "science": 70},
  "maria": {"maths" : 95, "English": 100, "Science": 90}
}
print(y["kelechi"]["maths"]) # this will print 90
```

```python
y["maria"]["maths"]= 100 #this will update the value of maths for maria to 100
print(y["maria"]["maths"]) # this will print 100
```

```python
y["Alex"] = {"maths": 80, "English": 70, "Science": 60}# this will add a new student to the dictionary
print(y)#this will print the entire dictionary
```

```python
print(y["kelechi"].get("maths"))
```
