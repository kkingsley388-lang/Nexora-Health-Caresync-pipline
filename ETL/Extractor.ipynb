```python
#making a copy of the DataFrame we wwant to work on for version control.
clean_patient_df = patient_df.copy(deep=True)
```

```python
clean_patient_df.head(2)
```

```python
# we want to perfrom transformation on all the columns. and this applies to all the columns with UUID. Trim the ID column to retain the last element when splitted by -. str.split is a string method
clean_patient_df["Id"] = clean_patient_df["Id"].str.split("-").str[4]
clean_patient_df.head(2)

#In Python, str[-1] outputs the last character of a string. For example, if s = "hello", then s[-1] returns "o"
```

```python
# we want to convert a data stored in "BIRTHDATE" column's from object to datetime format. handling errorrs by using "errors='coerce'" to coerce invalid formats to NaT
clean_patient_df["BIRTHDATE"] = pd.to_datetime(clean_patient_df["BIRTHDATE"], errors='coerce')
```

```python
#convert "DEATHDATE" to date time format. handling errors, like invalid data enteries in the column using "errors = corece" which replaces the invalid value with NaT(not a time).
clean_patient_df["DEATHDATE"] = pd.to_datetime(clean_patient_df["DEATHDATE"], errors="coerce" )
```

```python
clean_patient_df.head(2) #calling the dataframe we see what NaN has change to NaT
```

```python
#implement the PHI or PIP rules for our dataset. drop ssn, drivers, passport and suffix columns from the dataframe as they are not needed for further analysis.
clean_patient_df.drop(columns=["SSN", "DRIVERS", "PASSPORT","PREFIX"], inplace=True)# inplace = True. inplace: Makes changes in original Data Frame if True. 
```

```python
#strip off the numbers in the first, middle and last name columns to retain only the alphabetic characters. 
#clean_patient_df["FIRST"] = clean_patient_df["FIRST"].str.replace(r"\d+"," ", regex= True)
#"[0-9]+" → finds one or more numbers
#"" → replaces those numbers with nothing
#regex=True → tells pandas to treat "[0-9]+" as a regular expression
clean_patient_df["FIRST"] = clean_patient_df["FIRST"].str.replace("[0-9]+", "",regex= True)
clean_patient_df["MIDDLE"] = clean_patient_df["MIDDLE"].str.replace("[0-9]+", "",regex= True)
clean_patient_df["LAST"] = clean_patient_df["LAST"].str.replace("[0-9]+", "",regex= True)
```
